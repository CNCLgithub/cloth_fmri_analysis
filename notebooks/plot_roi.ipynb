```python
from os.path import join
import matplotlib.pyplot as plt
from nilearn import plotting
from cloth_fmri.config.config import CONFIG
from cloth_fmri.utils.fmri import get_roi_filename


def plot_roi(roi_type="tower"):
    for sub in CONFIG["subjects"]:
        fig = plt.figure(figsize=(12, 4))

        cur_roi_file = get_roi_filename(roi_type, sub)

        display = plotting.plot_glass_brain(
            cur_roi_file,
            display_mode="lyrz",
            title=f"{roi_type}_roi_sub{sub}",
            figure=fig,
        )

        for ax in fig.axes:
            for spine in ax.spines.values():
                spine.set_linewidth(0.1)

        plt.show()
        plt.close(fig)
```

```python
plot_roi(roi_type="tower")
```

```python
plot_roi(roi_type="v1")
```

```python
plot_roi(roi_type="loc")
```
