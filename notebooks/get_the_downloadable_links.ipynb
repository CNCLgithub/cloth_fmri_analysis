```python
import requests

def get_osf_files_with_names(node_id, folder_id):
    base_url = f"https://api.osf.io/v2/nodes/{node_id}/files/osfstorage/{folder_id}/"

    results = []
    url = base_url

    while url:
        r = requests.get(url)
        r.raise_for_status()
        data = r.json()

        for item in data["data"]:
            attr = item["attributes"]
            links = item["links"]

            # keep only files
            if attr.get("kind") != "file":
                continue

            file_name = attr["name"]

            # download link
            download_url = links.get("download")

            if not download_url:
                download_url = (
                    f"https://files.osf.io/v1/resources/{node_id}/providers/osfstorage/{item['id']}"
                )

            results.append({
                "name": file_name,
                "id": item["id"],
                "download": download_url
            })

        url = data["links"].get("next")

    return results


# ===== run =====
node_id = "kdmrg"
folder_id = "6a48a40b049df26716df9648"

files = get_osf_files_with_names(node_id, folder_id)

print("TOTAL FILES:", len(files))
print("-" * 60)

for f in files:
    print(f["name"])
    print(f["download"])
```
